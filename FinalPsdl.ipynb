```python
import cv2
import numpy as np
import time

# Capture video from webcam
cap = cv2.VideoCapture(0)

# Allow camera to warm up
print("Capturing background. Please stay still.")
time.sleep(3)

# Read background for a few seconds
for i in range(30):
    ret, background = cap.read()

# Flip the background (for mirror effect)
background = np.flip(background, axis=1)

print("Background captured. Now wear your cloak and enjoy the magic!")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Flip the frame for mirror image
    frame = np.flip(frame, axis=1)

    # Convert to HSV color space
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Define HSV range for cloak color (e.g., red cloak)
    lower_red1 = np.array([0, 120, 70])
    upper_red1 = np.array([10, 255, 255])
    lower_red2 = np.array([170, 120, 70])
    upper_red2 = np.array([180, 255, 255])

    # Create masks to detect red color
    mask1 = cv2.inRange(hsv, lower_red1, upper_red1)
    mask2 = cv2.inRange(hsv, lower_red2, upper_red2)
    cloak_mask = mask1 + mask2

    # Refine mask using morphology
    cloak_mask = cv2.morphologyEx(cloak_mask, cv2.MORPH_OPEN, np.ones((3, 3), np.uint8), iterations=2)
    cloak_mask = cv2.dilate(cloak_mask, np.ones((3, 3), np.uint8), iterations=1)

    # Invert the mask
    inverse_mask = cv2.bitwise_not(cloak_mask)

    # Segment out the cloak and replace with background
    cloak_area = cv2.bitwise_and(background, background, mask=cloak_mask)
    non_cloak_area = cv2.bitwise_and(frame, frame, mask=inverse_mask)
    final_output = cv2.addWeighted(cloak_area, 1, non_cloak_area, 1, 0)

    cv2.imshow("Invisibility Cloak - Harry Potter Style", final_output)

    # Break loop on 'q' key
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
